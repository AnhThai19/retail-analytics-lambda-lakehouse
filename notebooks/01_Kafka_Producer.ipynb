```python
import pandas as pd
import json
import time
import random
from kafka import KafkaProducer

# 1. Cấu hình Kafka kết nối NỘI BỘ
producer = KafkaProducer(
    bootstrap_servers=['kafka:9092'],
    value_serializer=lambda v: json.dumps(v).encode('utf-8')
)

TOPIC_NAME = 'retail_transactions'
CSV_FILE_PATH = 'OnlineRetail.csv'

def simulate_streaming():
    print(">>> Producer đang khởi động...")
    try:
        # Đọc theo chunksize để tiết kiệm RAM
        chunks = pd.read_csv(CSV_FILE_PATH, encoding='unicode_escape', chunksize=1000)
        
        total_sent = 0
        for chunk in chunks:
            records = chunk.to_dict(orient='records')
            
            # PHẢI CÓ VÒNG LẶP NÀY ĐỂ GỬI TỪNG DÒNG
            for record in records:
                producer.send(TOPIC_NAME, record)
                total_sent += 1
                
                # CHỈ SLEEP KHI CẦN DEMO. 
                # Nếu muốn chạy nhanh để lấy dữ liệu làm đồ án, hãy sửa thành 0.001 hoặc bỏ luôn sleep
                time.sleep(0.001) 
            
            # In log theo block để biết tiến độ
            print(f"--- Đã gửi tổng cộng: {total_sent:,} dòng ---")
            
    except Exception as e:
        print(f"Lỗi: {e}")

print("Hoàn thành thiết lập Producer.")

# Chạy hàm giả lập
if __name__ == "__main__":
    simulate_streaming()
```
